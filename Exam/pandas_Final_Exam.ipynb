# Use ipl-matches.csv

# 1 - how many matches has csk won in kolkata

```python
import pandas as pd

df=pd.DataFrame(pd.read_csv('ipl-matches.csv'))


print(df[['WinningTeam','City']][(df['WinningTeam']=='Chennai Super Kings') & (df['City']=='Kolkata')])
print(df[['WinningTeam','City']][(df['WinningTeam']=='Chennai Super Kings') & (df['City']=='Kolkata')].count())
# df[['WinningTeam','City']][( df['City']=='Kolkata')]
```

# 2 - toss winner is match winner in percentage

```python
import pandas as pd

df=pd.DataFrame(pd.read_csv('ipl-matches.csv'))
# print(df.head(20).to_string())

print(df[df['TossWinner'] == df['WinningTeam']].to_string())
# print(df['WinningTeam'])
```

# 4 - how many matches each team has played

```python
import pandas as pd

df=pd.DataFrame(pd.read_csv('ipl-matches.csv'))

t1=df['Team1'].value_counts()
t2=df['Team2'].value_counts()

total=t1+t2
print(total)
```
