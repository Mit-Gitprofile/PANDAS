# 2 - toss winner is match winner in percentage

```python
import pandas as pd

df=pd.DataFrame(pd.read_csv('ipl-matches.csv'))
# print(df.head(20).to_string())


toss_winner_match=df[df['TossWinner'] == df['WinningTeam']].shape[0]  # count the match toss winner is winning team
Total_matches=df.shape[0]                                             # count the total match played

per=(toss_winner_match/Total_matches)*100                             # find percentage
print(f"toss winner is match winner in percentage :",per)
```

# 2 -  find the batsman with max no of sixes

```python
import pandas as pd
df=pd.DataFrame(pd.read_csv('deliveries.csv'))

print(df[df['batsman_runs']==6]['batsman'].value_counts(ascending=False).head(1))
```

# 3 -  find batsman with most number of 4's and 6's in last 5 overs

```python
import pandas as pd
df=pd.DataFrame(pd.read_csv('deliveries.csv'))


last_overs=df[df['over']>=15]                                                               #find last five overs match
fourandsix=last_overs[(last_overs['batsman_runs']==6 )|( last_overs['batsman_runs']==4)]    #in last five overs six and four founds

print(fourandsix['batsman'].value_counts(ascending=False).head(5))              #convert to descending and head batsman
```

# 4 -  find V Kohli's record against all teams

```python
import pandas as pd
df=pd.DataFrame(pd.read_csv('deliveries.csv'))

virat=df[df['batsman']=='V Kohli']

print("\t   Virat kohli's record aginst all teams")
print("\t==============================================")
print(virat.groupby('bowling_team').agg(
    total_runs=('batsman_runs','sum'),
    total_balls=('batsman_runs','count')
).reset_index().to_string(index=False))
```

```python
import pandas as pd

# Function to find the highest score of any batsman
def highest_score(df, batsman_name):
    # Filter data for the given batsman
    batsman_df = df[df['batsman'] == batsman_name]
    
    # Calculate individual innings scores
    highest_score = batsman_df.groupby('match_id')['batsman_runs'].sum().max()
    
    return f"Highest score of {batsman_name}: {highest_score}"

# Sample Data
data = {
    'match_id': [1, 1, 1, 2, 2, 2, 3, 3],
    'batsman': ['A', 'A', 'A', 'B', 'B', 'B', 'A', 'A'],
    'batsman_runs': [4, 2, 6, 1, 4, 4, 6, 4]
}

df = pd.DataFrame(data)

# Example usage
print(highest_score(df, 'A'))  # Output: Highest score of A: 22
```

# 3 - Using Above DataFrame Filter Data with columns and row (10 opertions)

```python
data = {
    'ID':[101, 102, 103, 104],
    'Name':['John', 'Alice', 'Bob', 'Eve'],
    'Age':[23, 25, 22, 24],
    'TotalMarks':[89, 95, 88, 92],
    'SubjectCode':['MTH101', 'PHY202', 'CHM303', 'BIO404']
}
dict_df=pd.DataFrame(data)

print("--------------------------------------------")
print("         ORIGINAL DATAFRAME")
print("--------------------------------------------")

print(dict_df)
print("--------------------------------------------")

maxmark=dict_df['TotalMarks'].max()
minmark=dict_df['TotalMarks'].min()

maxname=dict_df[dict_df['TotalMarks']==maxmark]['Name']
minname=dict_df[dict_df['TotalMarks']==minmark]['Name']

Totalsum=dict_df['TotalMarks'].sum()
percentage=dict_df['Percentage'] = (dict_df['TotalMarks'] / Totalsum) * 100

print("\n-------------- MAX MARKS -----------------------\n")
print(f"Maximum marks:{maxmark} , Name :{maxname}")
print("\n-------------- MIN MARKS -----------------------\n")
print(f"minimum marks:{maxmark} , Name :{minname}")
print("\n-------------- PERCENTAGE ----------------------\n")
print(dict_df[['Name','TotalMarks','Percentage']])
print("\n--------------- BESIC INFO ---------------------\n")
print(dict_df.info())
print("\n--------------- SUMMARY -------------------------\n")
print(dict_df.describe())
print("\n--------------- SHAPE --------------------------\n")
print(dict_df.shape)
print("\n--------------- COLUMNS ------------------------\n")
print(dict_df.columns)
print("\n--------------- DATA TYPES----------------------\n")
print(dict_df.dtypes)
print("\n--------------- DIMENTION ----------------------\n")
print(dict_df.ndim)
print("\n--------------- ADD COLUMN ---------------------\n")
dict_df['Grade']=['A','B','C','D']
print(dict_df)
print("\n-------------------------------------------------\n")
```

# 4 - Apply GroupBy and Pivote method in any columns

```python
data = {
    'ID':[101, 102, 103, 104],
    'Name':['John', 'Alice', 'Bob', 'Eve'],
    'Age':[23, 25, 22, 24],
    'TotalMarks':[89, 95, 88, 92],
    'SubjectCode':['MTH101', 'PHY202', 'CHM303', 'BIO404']
}
dict_df=pd.DataFrame(data)
print("=============== GROUP BY =================\n")
print(dict_df.groupby('Name')['TotalMarks'].sum())

print("=============== PIVOTING ==================\n")
pivot=dict_df.pivot(index='ID',columns='Name',values='TotalMarks')
print(pivot)
```
